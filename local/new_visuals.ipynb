```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
```

```python
# Anon data
long = pd.read_csv('datasets/TEST2_anon_data.csv')
display(long)

# check how many repeatss
repeats = long.groupby(['clinical_id'], dropna=False, as_index=False).size()
repeat_counts = repeats['size'].value_counts().reset_index()
repeat_counts.rename(columns = {'size': 'number_visits', 'count': 'sample_count'}, inplace = True)
display(repeat_counts)
print(np.mean(repeats['size']))
```

```python
print(repeat_counts.sample_count.sum())
print(len(long.clinical_id.unique()))
```

```python
# Occurrences of each repeat visit

# Create the bar plot
fig = px.bar(
    repeat_counts, 
    x="number_visits", 
    y="sample_count", 
    labels={"sample_count": "Sample Count", "number_visits": "Number of Visits"}
)

# Improve layout
fig.update_traces(marker=dict(color="#A7C7E7"))
fig.update_layout(yaxis_categoryorder="total ascending", bargap=0.01, height = 600, width = 800)

# Show plot
fig.show()
```

```python
max_visits = long.groupby('clinical_id')['visit_month'].max().reset_index()
max_visits
```

```python
fig = px.bar(
    max_visits, 
    x="visit_month", 
    y="clinical_id", 
    orientation="h",
    labels={"clinical_id": "Sample ID", "visit_month": "Max Visit Month"}
)

fig.update_traces(marker=dict(color="#A7C7E7"))
fig.update_layout(yaxis_categoryorder="total ascending",bargap=0, height = 600, width = 800, yaxis=dict(showticklabels=False))

fig.show()
fig.write_html("max_visit_month_TUEPAC.html")
```
